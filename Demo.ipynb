### SkyScout demo

Try the fine-tuned detectors (YOLO11s baseline, YOLO26s newest YOLO, D-FINE fine-tuned) on drone photos and videos.

Click **Runtime > Run all**. It takes a few minutes the first time: it gets the code, installs the demo packages and downloads the trained models (about 80 MB). The app then shows up under the last cell, and a public `gradio.live` link is printed that also works in another tab or on a phone.

No dataset, Google Drive or training is needed. The full project (training and evaluation) is in `SkyScout.ipynb`.

#### 1. Get the code and install the demo packages

```python
import importlib
import os
import subprocess
import sys

REPO_URL = 'https://github.com/fnmoral/skyscout-small-object-detection-yolo-vs-dfine-visdrone.git'
PROJECT_DIR = os.getcwd() if os.path.exists('skyscout.py') else '/content/SkyScout'

if not os.path.exists(os.path.join(PROJECT_DIR, 'skyscout.py')):
    subprocess.run(['git', 'clone', '--depth', '1', REPO_URL, PROJECT_DIR], check=True)
elif PROJECT_DIR == '/content/SkyScout':
    # the copy on the Colab machine always follows the latest code on GitHub
    subprocess.run(['git', '-C', PROJECT_DIR, 'fetch', '--depth', '1', 'origin', 'main'], check=True)
    subprocess.run(['git', '-C', PROJECT_DIR, 'reset', '--hard', 'FETCH_HEAD'], check=True)
os.chdir(PROJECT_DIR)
# Demo packages only. keras-hub is installed without its dependencies so Colab keeps its own TensorFlow (no restart)
pip = [sys.executable, '-m', 'pip', 'install', '-q']
subprocess.run(pip + ['ultralytics', 'gradio>=6', 'imageio', 'imageio-ffmpeg', 'pycocotools', 'kagglehub',
                      'regex', 'rich', 'sentencepiece', 'tokenizers', 'keras>=3.15'], check=True)
subprocess.run(pip + ['--no-deps', 'keras-hub==0.33.0'], check=True)
sys.path.insert(0, PROJECT_DIR)

# running this cell again: stop the old app and load the new code
if 'skyscout' in sys.modules:
    for old_app in sys.modules['skyscout']._DEMO:
        old_app.close()
    sk = importlib.reload(sys.modules['skyscout'])
else:
    import skyscout as sk
version = subprocess.run(['git', 'log', '-1', '--format=%h %cd', '--date=short'], capture_output=True, text=True).stdout.strip()
print('Code version:', version, '| Profile:', sk.PROFILE, '| GPU:', sk.gpu_name())
```

#### 2. Download the trained models

```python
# Trained models from the GitHub release (checked against demo.models_sha256 before unzipping)
sk.fetch_models()
```

#### 3. Start the app

Pick a model, then try a sample photo or the sample clip. Turn on tiled inference (SAHI) for smaller objects.

```python
# Recommended: open the printed gradio.live link if the app below is too small
sk.launch_demo()
```
